```python
### Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)

### The copy.copy() function creates a shallow copy of the list. It creates a new outer list, but the inner lists are shared between a and b.

### b[0][1] = 99 changes the second element of the first inner list to 99.

### Since the inner lists are shared, the change also affects a.

### Therefore, the output is [[1, 99], [3, 4]].
```

```python
### Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)

### The copy.deepcopy() function creates a deep copy of the list, including its nested lists. Therefore, a and b have separate inner lists.
### b[0][1] = 99 changes the second element of the first inner list in b.
### The original list a remains unchanged because the nested lists are copied independently.
### Therefore, print(a) displays [[1, 2], [3, 4]].
```

```python
### Q38
import copy

a = [[10], [20]]
b = copy.deepcopy(a)
a[1].append(30)
print(b)

### The copy.deepcopy() function creates a deep copy of the list, including its inner lists. Therefore, a and b have independent nested lists.
### b = copy.deepcopy(a) creates a separate copy of a.
### a[1].append(30) adds 30 to the second inner list of a.
### The change does not affect b because its inner lists are independent.
### Therefore, the output is [[10], [20]].
```

```python
### Q44
def test(*args):
    print(args)

test(1, 2)

### *args allows a function to accept any number of positional arguments.

### When test(1, 2) is called, the values 1 and 2 are collected into a tuple named args.

### The print(args) statement displays the tuple (1, 2).
```

```python
### Q45
def test(*args):
    print(type(args))

test("a", "b", "c")

### *args allows a function to accept any number of positional arguments.

### Python collects all the arguments ("a", "b", and "c") into a tuple named args.

### The type() function returns the data type of args, which is tuple.
```

```python
### Q46
def test(**kwargs):
    print(kwargs)

test(name="Ram", age=25)

### **kwargs allows a function to accept any number of keyword arguments.

### Python collects the keyword arguments into a dictionary named kwargs.

### Here, name="Ram" and age=25 are stored as key-value pairs in the dictionary.

### The print(kwargs) statement displays the dictionary.
```

```python
### Q47
def test(**kwargs):
    print(type(kwargs))

test(a=10, b=20)

### **kwargs allows a function to accept any number of keyword arguments.
### Python collects the keyword arguments a=10 and b=20 into a dictionary named kwargs.
### The type() function checks the data type of kwargs.
### Therefore, the output is <class 'dict'>.
```
