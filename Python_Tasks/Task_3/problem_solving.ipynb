### Q1. Find the Frequency of Each Word in a Sentence


START
  freq = empty dictionary
  words = split sentence by spaces
  FOR each word in words
    IF word in freq THEN freq[word] = freq[word] + 1
    ELSE freq[word] = 1
  RETURN freq
END

```python
def word_frequency(sentence):
    freq = {}
    for word in sentence.split():
        freq[word] = freq.get(word, 0) + 1
    return freq

print(word_frequency("python is easy and python is powerful"))
# Expected: {'python': 2, 'is': 2, 'easy': 1, 'and': 1, 'powerful': 1}
print(word_frequency("to be or not to be"))
# {'to': 2, 'be': 2, 'or': 1, 'not': 1}
```

### Q2. Find the Frequency of Each Character in a Word
Q2. CHARACTER FREQUENCY
START
  freq = empty dictionary
  FOR each ch in word
    IF ch in freq THEN freq[ch] = freq[ch] + 1
    ELSE freq[ch] = 1
  RETURN freq
END

```python
def char_frequency(word):
    freq = {}
    for ch in word:
        freq[ch] = freq.get(ch, 0) + 1
    return freq

print(char_frequency("banana"))   # Expected: {'b': 1, 'a': 3, 'n': 2}
print(char_frequency("hello"))    # {'h': 1, 'e': 1, 'l': 2, 'o': 1}
```

### Q3. Find the First Non-Repeating Character in a Word
START
  freq = character frequency of word
  FOR each ch in word (left to right)
    IF freq[ch] == 1 THEN RETURN ch
  RETURN None
END

```python
def first_non_repeating_char(word):
    freq = char_frequency(word)
    for ch in word:
        if freq[ch] == 1:
            return ch
    return None

print(first_non_repeating_char("swiss"))  # Expected: w
print(first_non_repeating_char("aabb"))   # None
```

### Q4. Check Whether Two Strings Are Anagrams
Q4. ANAGRAM CHECK
START
  freq1 = character frequency of s1
  freq2 = character frequency of s2
  IF freq1 == freq2 THEN RETURN True
  ELSE RETURN False
END

```python
def are_anagrams(s1, s2):
    return char_frequency(s1) == char_frequency(s2)

print(are_anagrams("listen", "silent"))  # Expected: True
print(are_anagrams("hello", "world"))    # False
```

### Q5. Find the First Repeating Character in a String

START
  seen = empty set
  FOR each ch in text (left to right)
    IF ch in seen THEN RETURN ch
    ADD ch to seen
  RETURN None
END

```python
def first_repeating_char(text):
    seen = set()
    for ch in text:
        if ch in seen:
            return ch
        seen.add(ch)
    return None

print(first_repeating_char("programming"))  # Expected: r
print(first_repeating_char("abc"))          # None
```

### Q6. Move All Zeros to the End of a List

START
  non_zeros = empty list
  FOR each x in nums
    IF x != 0 THEN append x to non_zeros
  zero_count = length of nums - length of non_zeros
  RETURN non_zeros + (zero_count zeros)
END

```python
def move_zeros_to_end(nums):
    non_zeros = [x for x in nums if x != 0]
    zeros = [0] * (len(nums) - len(non_zeros))
    return non_zeros + zeros

print(move_zeros_to_end([0, 3, 0, 1, 2, 0, 4]))  # Expected: [3, 1, 2, 4, 0, 0, 0]
print(move_zeros_to_end([0, 0, 5]))              # [5, 0, 0]
```

### Q7. Find the Second Highest Distinct Number in a List

START
  distinct = unique values of nums, sorted high to low
  IF length of distinct < 2 THEN RETURN None
  RETURN distinct[1]
END

```python
def second_highest(nums):
    distinct = sorted(set(nums), reverse=True)
    if len(distinct) < 2:
        return None
    return distinct[1]

print(second_highest([10, 20, 40, 30, 40, 20]))  # Expected: 30
print(second_highest([5, 5, 5]))                 # None
```

### Q8. Remove Duplicates Without Changing the Original Order

START
  seen = empty set
  result = empty list
  FOR each x in nums
    IF x not in seen THEN
      add x to seen
      append x to result
  RETURN result
END

```python
def remove_duplicates(nums):
    seen = set()
    result = []
    for x in nums:
        if x not in seen:
            seen.add(x)
            result.append(x)
    return result

print(remove_duplicates([4, 2, 4, 5, 2, 7, 5]))  # Expected: [4, 2, 5, 7]
print(remove_duplicates([1, 1, 1]))              # [1]
```

### Q9. Find the Missing Number from 1 to N

START
  expected = N * (N + 1) / 2
  actual = sum of nums
  RETURN expected - actual
END

```python
def find_missing_number(nums, n):
    return n * (n + 1) // 2 - sum(nums)

print(find_missing_number([1, 2, 4, 5, 6], 6))  # Expected: 3
print(find_missing_number([2, 3, 4, 5], 5))     # 1
```

### Q10. Find Two Numbers Whose Sum Equals a Target

START
  seen = empty set
  FOR each x in nums (left to right)
    need = target - x
    IF need in seen THEN RETURN [need, x]
    ADD x to seen
  RETURN empty list
END

```python
def find_pair_with_sum(nums, target):
    seen = set()
    for x in nums:
        need = target - x
        if need in seen:
            return [need, x]
        seen.add(x)
    return []

print(find_pair_with_sum([2, 7, 11, 15], 9))  # Expected: [2, 7]
print(find_pair_with_sum([1, 2, 3], 10))      # []
```

### Q11. Find the Number Closest to Zero

START
  best = first element of nums
  FOR each x in nums
    IF abs(x) < abs(best) THEN best = x
    ELSE IF abs(x) == abs(best) AND x > best THEN best = x
  RETURN best
END

```python
def closest_to_zero(nums):
    return min(nums, key=lambda x: (abs(x), -x))

print(closest_to_zero([-8, -3, 2, 5, -2]))  # Expected: 2
print(closest_to_zero([-4, 4, 7]))          # 4
```

### Q12. Find Common Elements Between Two Lists

START
  set2 = set of list2
  result = empty list
  FOR each x in list1 (in order)
    IF x in set2 AND x not in result THEN append x to result
  RETURN result
END

```python
def common_elements(list1, list2):
    set2 = set(list2)
    result = []
    for x in list1:
        if x in set2 and x not in result:
            result.append(x)
    return result

print(common_elements([1, 2, 3, 4, 5], [3, 4, 5, 6, 7]))  # Expected: [3, 4, 5]
print(common_elements([1, 1, 2], [1, 3]))                 # [1]
```

### Q13. Rotate a List to the Right by K Positions

START
  IF nums is empty THEN RETURN empty list
  k = k mod length of nums
  IF k == 0 THEN RETURN copy of nums
  RETURN (last k elements of nums) + (all elements before them)
END

```python
def rotate_right(nums, k):
    if not nums:
        return []
    k = k % len(nums)
    if k == 0:
        return nums[:]
    return nums[-k:] + nums[:-k]

print(rotate_right([1, 2, 3, 4, 5], 2))  # Expected: [4, 5, 1, 2, 3]
print(rotate_right([1, 2, 3], 5))        # [2, 3, 1]
```

### Q14. Find the Maximum Sum of Two Consecutive Elements
START
  best = nums[0] + nums[1]
  FOR i from 1 to length of nums - 2
    current = nums[i] + nums[i+1]
    IF current > best THEN best = current
  RETURN best
END

```python
def max_consecutive_sum(nums):
    best = nums[0] + nums[1]
    for i in range(1, len(nums) - 1):
        best = max(best, nums[i] + nums[i + 1])
    return best

print(max_consecutive_sum([4, 7, 2, 9, 5]))  # Expected: 14
print(max_consecutive_sum([-5, -1, -3]))     # -4
```

### Q15. Find All Pairs of Numbers Whose Sum Equals a Target

START
  pairs = empty set
  FOR i from 0 to length of nums - 1
    FOR j from i + 1 to length of nums - 1
      IF nums[i] + nums[j] == target THEN
        add (smaller, larger) of the two numbers to pairs
  result = pairs converted to list of lists, sorted
  RETURN result
END

```python
def find_all_pairs(nums, target):
    pairs = set()
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                pairs.add(tuple(sorted((nums[i], nums[j]))))
    return [list(p) for p in sorted(pairs)]

print(find_all_pairs([2, 4, 3, 5, 7, 8, 1], 9))  # Expected: [[1, 8], [2, 7], [4, 5]]
print(find_all_pairs([3, 3, 3], 6))              # [[3, 3]]
```
